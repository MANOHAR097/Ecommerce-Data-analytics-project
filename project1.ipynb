Null values replace with values

```python
# Fill Discount_Percent with mean
df["Discount_Percent"] = df["Discount_Percent"].fillna(
    df["Discount_Percent"].mean())
```

```python
# Fill Customer_Rating with median
df["Customer_Rating"] = df["Customer_Rating"].fillna(df["Customer_Rating"].median())
```

```python
# Fill Shipping_City with most frequent city
df["Shipping_City"] = df["Shipping_City"].fillna(df["Shipping_City"].mode()[0])
```

```python
print(df.isnull().sum())
```

Convert Order_Date to proper Date

```python
df["Order_Date"]=pd.to_datetime(df["Order_Date"], format='mixed')
print(df["Order_Date"].dtype)
```

Change negative values into positive values

```python
df['Quantity'].replace(-1, 0, inplace=True)
```

Remove rows with invalid quantities

```python
df = df[df['Quantity'] > 0]
```

```python
df.loc[:, "Payment_Method"] = df["Payment_Method"].replace({
    "upi": "UPI",
    "U.P.I": "UPI",
    "Credit_card": "Credit Card",
    "Credit Card": "Credit Card",
    "Cash on Delivery": "Cash on Delivery",
    "cod": "Cash on Delivery",
    "NetBanking": "Net Banking",
    "Net Banking": "Net Banking",
    "Pay Pal": "Pay Pal",
    "paypal": "Pay Pal",
})
```

```python
df.loc[:, "Country"] = df["Country"].replace({
    "Australia": "Australia",
    "AU": "Australia",
    "India": "India",
    "IN": "India",
    "United States": "United States",
    "US": "United States",
    "USA": "United States",
    "U.S.A.": "United States",
    "United Kingdom": "United Kingdom",
    "UK": "United Kingdom",
    "U.K.": "United Kingdom",
    "Germany": "Germany",
    "DE": "Germany",
    "Canada": "Canada",
    "CA": "Canada",
    "U.A.E.": "United Arab Emirates",
    "UAE": "United Arab Emirates",})
```

```python
df.to_csv('ecommerce_retail_transactions_cleaned.csv', index=False)
```
